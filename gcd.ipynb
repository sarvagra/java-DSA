## Computing GCD 


### Using normal loop approach, which has high time and space complexity - fails for large numbers

```python
# loop based approach 
def gcd(m,n):
    a=max(m,n)
    b=min(m,n)
    for i in range(1,b+1):
        if a%i==0 and b%i==0:
            x=i
    return x
print(gcd(6,15))
```

### Recursive approach using mathematical observations, which reduces space and time complexity upto some extent

```python
# recursive approach
def gcd(m,n):
    
    a=min(m,n)
    b=max(m,n)
    if (a==0 and b==0):
        return 0
    elif(a==0 or b==0):
        return b
    elif(b%a==0):
        return a
    elif a!=b:
        return gcd(a,b-a)
    else: 
        return a
        
print(gcd(6,0)) #print(gcd(2,99997))
# suppose using for print(gcd(2,99997)), itll crash - which leads us to using Euclid's algorithm
```

### Euclids algorithm(again mathematical obsservation), to reduce recursive calls 

```python
def gcd(m,n):
    a = max(m,n)
    b = min(m,n)

    if b == 0:
        return a
    elif a%b == 0:
        return b
    else:
        return gcd(b, a%b)

print(gcd(6,0))      # 6
print(gcd(2,99997))  # 1
```
